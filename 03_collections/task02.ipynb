```python
keys = tuple("abcdefghijklmnopqrstuvwxyz")
input_alphabet = dict.fromkeys(keys, 0)
input_tangos = []

while True:
    input_tango = input("英単語を入力してください：")
    if input_tango == "":
        break
    elif input_tango.isalpha():
        input_tangos.append(input_tango.lower())
    else:
        print("エラー: アルファベットを入力してください。")

print(f"入力した英単語：{sorted(input_tangos)}")

for tango in input_tangos:
    for input_char in tango:
        input_alphabet[input_char] += 1

for key in input_alphabet:
    if input_alphabet[key] > 0:
        print(f"{key}が{input_alphabet[key]}個ありました")
```
