```python
grades = [85, 90, 78, 92, 88]
sumgrades = 0
for i in grades:
   sumgrades += i
average = sumgrades / len(grades)
print(f"合計: {sumgrades}")
print(f"平均点: {average}")

grades_sort = grades.copy()
grades_sort.sort()
print(f"最高点: {grades_sort[-1]}")  
print(f"最低点: {grades_sort[0]}")

over_average = [i for i in grades if average < i]
print(f"平均以上の学生数: {len(over_average)}")
```
